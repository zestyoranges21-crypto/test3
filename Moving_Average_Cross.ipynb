```python
from datetime import timedelta
import numpy as np
import pandas as pd
```

```python
time = []
price = []

with open("log_data_500.txt", "r") as f:
    for line in f:
        parts = line.strip().split(", ")
        t = parts[0].split("timestamp: ")[1]
        p = float(parts[1].split("price: ")[1])
        time.append(t)
        price.append(p)

# Convert to pandas DataFrame (optional)
import pandas as pd
df = pd.DataFrame({"timestamp": pd.to_datetime(time), "price": price})
print(df.head())
```

```python
time = pd.to_datetime(time)
```

```python

def avg(Mat):
    return np.mean(np.array(Mat))
for i in range(len(price)):
    if time[i] - time[0] > timedelta(minutes = 10):
        initial_time = i
        break

def MA_one(time, price, i):
    for t in range(i+1):
        if time[i] - time[i-t] > timedelta(minutes = 1):
            lower = i-t + 1
            break
    return float(avg(price[lower:i+1])) 

def MA_ten(time, price, i):
    for t in range(i+1):
        if time[i] - time[i-t] > timedelta(minutes = 10):
            lower = i-t + 1
            break
    return float(avg(price[lower:i+1]))

def sign(x):
    return 1 if x >= 0 else -1

i = initial_time + 1
position = 0
profit = 0
while i < len(price):
    if (MA_one(time, price, i) - MA_ten(time, price, i)) * (MA_one(time, price, i-1) - MA_ten(time, price, i-1)) < 0:
        position = sign(MA_one(time, price, i) - MA_ten(time, price, i))
        profit -= sign(MA_one(time, price, i) - MA_ten(time, price, i))*price[i]
    i += 1
profit += position * price[-1]
```
