```python
def bfs(initial, connections):
    queue = [initial]
    order = []
    while queue:
        value = queue.pop(0)
        order.append(value)
        visited[value] = True
        for i in range(0,len(connections)):
            if connections[value][i] == 1 and i not in order:
                queue.append(i)
    return order
```
